# Algoritmo dos Vizinhos Mais Próximos (k-NN)

O algoritmo dos **Vizinhos Mais Próximos (k-NN)** é um método usado para classificação e regressão. É um dos algoritmos de aprendizado de máquina mais simples e intuitivos.

Basicamente, o k-NN funciona encontrando os `k` pontos de dados mais próximos (vizinhos) do novo ponto a ser classificado ou ter seu valor previsto. A decisão sobre a classe ou o valor do novo ponto é tomada com base nesses `k` vizinhos. Em um problema de classificação, o novo ponto é atribuído à classe que é mais comum entre seus `k` vizinhos mais próximos (voto da maioria).

Uma premissa fundamental do k-NN é a **hipótese de que objetos relacionados estão próximos no espaço vetorial**. Isso significa que, se dois pontos de dados são semelhantes (têm características parecidas), eles tendem a estar próximos um do outro em um espaço multidimensional. A proximidade é geralmente medida usando funções de distância, como a distância Euclidiana.

O k-NN é considerado um **algoritmo "preguiçoso" (lazy learner)**. Diferente de outros algoritmos, ele não aprende um modelo compacto dos dados de treinamento durante a fase de treinamento. Em vez disso, ele:

*   **Memoriza os objetos de treinamento**: Durante a fase de treinamento, o k-NN simplesmente armazena todo o conjunto de dados de treinamento. Não há um processo explícito de "aprendizagem" para criar um modelo generalizado.
*   **Realiza computação apenas no momento da previsão**: A maior parte do trabalho computacional acontece apenas quando uma nova previsão é solicitada. Neste momento, o algoritmo calcula as distâncias entre o novo ponto e todos os pontos de treinamento para encontrar os `k` vizinhos mais próximos.

Essa natureza "preguiçosa" tem suas vantagens e desvantagens. Uma vantagem é a sua simplicidade e capacidade de se adaptar bem a dados complexos sem a necessidade de suposições sobre a distribuição dos dados. No entanto, o método pode ser computacionalmente caro para grandes conjuntos de dados, pois cada previsão requer uma busca através de todo o conjunto de treinamento.

## O caso do vizinho mais próximo (1-NN)

Quando escolhemos `k = 1`, o algoritmo k-NN se torna o **Vizinho Mais Próximo (1-NN)**. Neste cenário simplificado, a decisão de classificação para um novo ponto é baseada unicamente na instância de treinamento mais próxima.

> Algoritmo do 1-vizinho mais próximo
>
> **Entrada**:
> - Um conjunto de treinamento $\textbf{D}$ com $n$ instâncias, cada uma associada a uma classe.
> - Uma instância de teste $\textbf{t}$ a ser classificada.
> - Uma função $d(\textbf{x}_a, \textbf{x}_b)$ que calcula a distância entre duas instâncias $\textbf{x}_a$ e $\textbf{x}_b$.
>
> **Saída**:
> - A classe atribuída à instância $\textbf{t}$.
>
> $d_{min} \gets +\infty $
>
> Para cada instância $\textbf{x}_i$ em $\textbf{D}$:
>
> &ensp; &ensp;$d_{i} \gets d(\textbf{x}_i, \textbf{t}) $
>
> &ensp; &ensp;Se $d_i < d_{min}$:
>
> &ensp; &ensp;&ensp; &ensp;$d_{min} \gets d_{i}$
>
> &ensp; &ensp;&ensp; &ensp;$c_{min} \gets $ classe associada a $\textbf{x}_i$
>
> Retorne $c_{min}$

### Exemplo: classificação da Iris com 1-NN

Neste exemplo, usaremos o dataset da Iris para demonstrar como o algoritmo k-NN, com $k=1$, pode ser aplicado a um problema de classificação. O conjunto de dados Iris contém medições de quatro características (comprimento e largura da sépala, comprimento e largura da pétala) para 150 flores de três espécies diferentes de Iris.

O primeiro passo é carregar o dataset Iris. Em seguida, dividiremos os dados em conjuntos de treinamento e teste para avaliar o desempenho do nosso modelo.

````python
import numpy as np
from sklearn.datasets import load_iris

# Carregando o conjunto da Iris
iris = load_iris()
X = iris.data   # Características
y = iris.target # Rótulos (classes)

print(f"Shape das características (X): {X.shape}")
print(f"Shape dos rótulos (y): {y.shape}")
print(f"Nomes das classes: {iris.target_names}")
````

````python
# Dividindo os dados em conjuntos de treinamento e teste
from sklearn.model_selection import train_test_split

# Usaremos 5 instâncias de teste por classe, totalizando 15 instâncias de teste
test_size_absolute = 15

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=test_size_absolute, # tamanho do conjunto de teste
    stratify=y,                   # conjunto que será usado para guiar a estratificação. Mantém a proporção da ocorrência de classes no conjunto.
    random_state=42               # semente de aleatoriedade. Garante os memos resultados entre execuções distintas.
)

print(f"Shape de X_train: {X_train.shape}") # instâncias de treinamento
print(f"Shape de X_test: {X_test.shape}")   # instâncias de teste
print(f"Shape de y_train: {y_train.shape}") # classes das instâncias de treinamento
print(f"Shape de y_test: {y_test.shape}")   # classes das instâncias de teste
````

Agora, inicializaremos o classificador k-NN. Para este exemplo, definiremos `n_neighbors=1`, o que corresponde ao caso do Vizinho Mais Próximo. Utilizaremos a implementação do algoritmo k-NN provida pelo scikit-learn.

````python
# Inicializando o classificador k-NN com k=1
from sklearn.neighbors import KNeighborsClassifier

knn_classifier = KNeighborsClassifier(n_neighbors=1)
````

Em seguida, treinamos o modelo usando os dados de treinamento. Na natureza 'preguiçosa' do k-NN, isso significa que o modelo simplesmente 'memoriza' os dados de treinamento.

````python
# Treinando o modelo
knn_classifier.fit(X_train, y_train)
print('Treinamento finalizado com sucesso!')
````

Após o treinamento, avaliamos o desempenho do modelo comparando as previsões com os rótulos reais do conjunto de teste. A métrica de acurácia (accuracy) nos dirá a proporção de previsões corretas.

````python
from sklearn.metrics import accuracy_score

# Obtendo as respostas do classificador para as instâncias de teste
y_pred = knn_classifier.predict(X_test)

# Calculando a acurácia total por meio da função accuracy_score do scikit-learn
overall_accuracy = accuracy_score(
    y_test, # respostas corretas
    y_pred  # respostas/previsões geradas pelo classificador
)
print(f"Acurácia total: {overall_accuracy:.2f}")

# Calculando a acurácia por classe
unique_classes = np.unique(y_test) # Obtendo as classes únicas
for class_label in unique_classes:
    # Filtrando os rótulos reais e as previsões para a classe atual
    class_indices = (y_test == class_label)
    y_test_class = y_test[class_indices]
    y_pred_class = y_pred[class_indices]

    # Calculando a acurácia para a classe atual
    accuracy_class = accuracy_score(y_test_class, y_pred_class)

    # Obtendo o nome da classe para melhor legibilidade
    class_name = iris.target_names[class_label]
    print(f"  Classe '{class_name}' ({class_label}): {accuracy_class:.2f}")
````

## Algoritmo k-NN

Uma utilização comum do algoritmo é adotar valores de `k` maiores do que 1. Em problemas de classificação, cada vizinho mais próximo conta como um voto na respectiva classe. Ao final, os votos são agregados por classe e a instância de teste é associada à classe com maior número de votos.

Segue um pseudocódigo para o algoritmo k-NN voltado para problemas de classificação:

```pseudocode
Função k_NN_Classificacao(ConjuntoDeTreinamento, NovoPonto, k):
  // ConjuntoDeTreinamento: Um conjunto de pontos de dados com rótulos de classe conhecidos
  // NovoPonto: O ponto para o qual queremos prever a classe
  // k: O número de vizinhos mais próximos a considerar

  1. Inicialize uma lista vazia de DistânciasEVizinhos.

  2. Para cada PontoDeTreinamento no ConjuntoDeTreinamento:
     a. Calcule a Distância(NovoPonto, PontoDeTreinamento).
     b. Adicione (Distância, PontoDeTreinamento) à lista DistânciasEVizinhos.

  3. Ordene a lista DistânciasEVizinhos em ordem crescente pela Distância.

  4. Selecione os primeiros k itens da lista ordenada. Estes são os k_VizinhosMaisProximos.

  5. Obtenha os rótulos (classes) de cada vizinho em k_VizinhosMaisProximos.

  6. Encontre a ClasseMaisComum entre esses rótulos (usando votação majoritária).

  7. Retorne a ClasseMaisComum como a previsão para NovoPonto.
```

Explicação do pseudocódigo:

1.  **Inicialização**: Prepara uma lista para armazenar as distâncias e os respectivos pontos de treinamento.

2.  **Cálculo de Distâncias**: Para cada ponto no conjunto de treinamento, o algoritmo calcula a distância entre esse ponto e o `NovoPonto`. A escolha da métrica de distância (Euclidiana, Manhattan, etc.) é crucial aqui e depende da natureza dos dados.

3.  **Ordenação**: Todas as distâncias calculadas são ordenadas, geralmente da menor para a maior. Isso nos permite identificar quais pontos estão mais próximos do `NovoPonto`.

4.  **Seleção dos `k` Vizinhos**: Os `k` primeiros pontos da lista ordenada são os vizinhos mais próximos. Estes são os pontos que terão influência na previsão do `NovoPonto`.

5.  **Obtenção dos Rótulos**: O algoritmo verifica a classe de cada um dos `k` vizinhos selecionados.

6.  **Votação Majoritária**: Para classificar o `NovoPonto`, o algoritmo atribui a ele a classe que aparece com maior frequência entre os `k` vizinhos (voto majoritário).

7.  **Retorno da Previsão**: A classe mais comum é retornada como a previsão para o `NovoPonto`.

A escolha do valor adequado de `k` pode não ser trivial e depende do problema em questão. Em problemas de classificação com duas classes costuma-se evitar `k` com valores pares para evitar empates.

- Estimar `k` por validação cruzada;
- Associar um peso para a contribução de cada vizinho. O peso deve ser inversamente proporcional à distância ao ponto de teste.

### Exemplo: classificação da Iris com 3-NN e 5-NN

Neste exemplo, usaremos o dataset da Iris novamente para demonstrar o uso do algoritmo k-NN para $k=3$ e $k=5$.

````python
def treinar_e_avaliar_knn(k):
    knn_classifier = KNeighborsClassifier(n_neighbors=k)
    knn_classifier.fit(X_train, y_train)

    y_pred = knn_classifier.predict(X_test)
    overall_accuracy = accuracy_score(
        y_test, # respostas corretas
        y_pred  # respostas/previsões geradas pelo classificador
    )
    print(f'## Desempenho do classificador {k}-NN:')
    print(f"  Acurácia total: {overall_accuracy:.2f}")

    # Calculando a acurácia por classe
    unique_classes = np.unique(y_test) # Obtendo as classes únicas
    for class_label in unique_classes:
        # Filtrando os rótulos reais e as previsões para a classe atual
        class_indices = (y_test == class_label)
        y_test_class = y_test[class_indices]
        y_pred_class = y_pred[class_indices]

        # Calculando a acurácia para a classe atual
        accuracy_class = accuracy_score(y_test_class, y_pred_class)

        # Obtendo o nome da classe para melhor legibilidade
        class_name = iris.target_names[class_label]
        print(f"  Classe '{class_name}' ({class_label}): {accuracy_class:.2f}")

treinar_e_avaliar_knn(3)
treinar_e_avaliar_knn(5)
````

## Vantagens e Desvantagens do Algoritmo k-NN

O algoritmo k-NN, apesar de sua simplicidade, possui características que o tornam vantajoso ou desvantajoso dependendo do contexto do problema.

Vantagens:
- **Simplicidade e intuitividade**: É um dos algoritmos de aprendizado de máquina mais fáceis de entender e implementar. Sua lógica é direta: a similaridade entre pontos de dados é a base para as decisões.
- **Não-paramétrico**: O k-NN não faz suposições sobre a distribuição subjacente dos dados (ao contrário de algoritmos paramétricos como a regressão linear ou regressão logística, que assumem normalidade ou linearidade). Isso o torna flexível e eficaz para dados com distribuições complexas ou irregulares.
- **Flexibilidade para diferentes tipos de dados**: Pode ser aplicado tanto a problemas de classificação quanto de regressão e funciona bem com dados numéricos e categóricos (desde que uma métrica de distância apropriada seja usada).
- **Aprendizado lazy (Preguiçoso)**: Como um *lazy learner*, ele não gasta tempo construindo um modelo complexo durante a fase de treinamento, apenas armazena os dados. Isso significa que o treinamento é muito rápido, pois não há um processo de "aprendizagem" explícito.
- **Adaptação a novas informações**: Como o modelo não é construído explicitamente, a adição de novos dados de treinamento é trivial e não requer o retreinamento completo do modelo (apenas o conjunto de dados é atualizado).

Desvantagens:

- **Custo computacional elevado em tempo de predição**: A principal desvantagem é o alto custo computacional na fase de previsão, especialmente para grandes conjuntos de dados. Para classificar um único novo ponto, o algoritmo precisa calcular a distância desse ponto para *todas* as instâncias no conjunto de treinamento.
- **Sensibilidade à escala das características**: O k-NN é sensível à escala dos dados. Se as características tiverem escalas muito diferentes, aquelas com escalas maiores podem dominar o cálculo da distância, distorcendo os resultados. É fundamental realizar a normalização ou padronização dos dados antes de aplicar o k-NN.
- **Sensibilidade a ruídos**: Como todo algoritmo baseado em distâncias, o k-NN é afetado pela presença de atributos redundantes ou irrelevantes.
- **Maldição da dimensionalidade**: Em espaços de alta dimensionalidade (muitas características), a noção de "proximidade" se torna menos significativa. A distância entre os pontos tende a se tornar mais uniforme, tornando difícil encontrar vizinhos verdadeiramente próximos, o que pode degradar o desempenho do modelo.
- **Escolha do valor de `k`**: A escolha do número ideal de vizinhos (`k`) é crucial e muitas vezes empírica. Um `k` muito pequeno pode ser sensível a ruídos, enquanto um `k` muito grande pode suavizar demais a decisão, perdendo detalhes importantes e fazendo com que os vizinhos não sejam tão "locais".
