# Avaliação de modelos de classificação

A avaliação do desempenho de modelos preditivos é uma etapa fundamental do Aprendizado de Máquina. No contexto acadêmico e profissional, construir um modelo capaz de aprender com dados históricos é apenas o primeiro passo; o verdadeiro desafio reside em mensurar o quão bem esse modelo se comportará ao enfrentar dados reais e inéditos, ou seja, sua capacidade de generalização. Sem um processo criterioso de avaliação, há o risco de implantar algoritmos ineficientes, enviesados ou potencialmente nocivos em aplicações críticas como diagnósticos médicos, geração de alertas e veículos autônomos.

## Dilema  Viés-Variância (*Bias-Variance Trade-off*)

A principal razão para avaliar um modelo é verificar se ele foi capaz de extrair padrões subjacentes relevantes a partir dos dados de treinamento, em vez de simplesmente memorizá-los. Esse comportamento é governado pelo **compromisso entre viés e variância** (*bias-variance trade-off*), um conceito central em aprendizado estatístico.

O **viés** (*bias*) mede o erro introduzido pelas suposições simplistas do algoritmo. Um modelo com alto viés é pouco flexível e tende a ignorar as relações complexas dos dados, resultando em subajuste (*underfitting*), onde o desempenho é fraco tanto no treinamento quanto no teste.

Por outro lado, um modelo com alta **variância** (*variance*) é extremamente flexível e tende a a seguir as mínimas variações e ruídos presentes nos dados. Um modelo com alta variância decorará as particularidades dos dados de treino, resultando em sobreajuste (*overfitting*), apresentando excelente desempenho durante o treino, mas falhando ao classificar amostras inéditas. O objetivo da avaliação é identificar o ponto ideal de equilíbrio em que o erro total de generalização seja minimizado.

A importância da avaliação torna-se ainda mais evidente em *datasets* desbalanceados, onde uma classe é extremamente severa ou rara em comparação às outras. Considere um problema médico em que apenas 2% dos pacientes avaliados possuem uma doença rara (20 doentes em 1.000 indivíduos). Um modelo ingênuo que simplesmente preveja que "nenhum paciente está doente" alcançará uma acurácia impressionante de 98\%. No entanto, esse classificador é inútil do ponto de vista prático, pois possui taxa de acerto nula para a classe de maior interesse clínico. Nesses cenários, técnicas de avaliação adequadas e métricas robustas são indispensáveis para revelar as verdadeiras deficiências do sistema.

## Divisão de Dados

Um erro conceitual clássico é avaliar o modelo utilizando exatamente os mesmos dados em que ele foi treinado. Essa prática é chamada de avaliação por resubstituição e produz uma estimativa excessivamente otimista e ilusória do desempenho do classificador. Avaliar por resubstituição equivale a aplicar uma prova a alunos contendo exatamente as mesmas perguntas e respostas fornecidas no gabarito durante a aula: a nota máxima refletirá memorização perfeita, não aprendizado real.

Para contornar esse problema e obter estimativas não enviesadas, a prática padrão consiste em particionar os dados em três conjuntos distintos e disjuntos: treinamento, validação e teste.

O conjunto de **Treinamento** (*Training Set*) é a parcela de dados fornecida ao algoritmo para o ajuste de seus parâmetros internos.

> **Parâmetros** de um modelo de aprendizado de máquina são variáveis internas cujos valores são ajustados automaticamente pelo algoritmo durante o treinamento, a partir dos dados de entrada. Eles determinam como o modelo transforma dados e faz previsões. Exemplos: pesos de uma rede neural, nós de uma árvore de decisão e os vetores de suporte em um modelo SVM.

O conjunto de **Validação** (*Validation Set*) é reservado para a etapa de ajuste de hiperparâmetros (por exemplo, a profundidade máxima de uma árvore ou a taxa de aprendizado) e seleção de modelos. Ele atua como um intermediário para monitorar a ocorrência de *overfitting* sem contaminar a avaliação final.

> **Hiperparâmetros** são variáveis de configuração definidas antes do início do treinamento de um modelo de aprendizado de máquina. Ao contrário dos parâmetros, que o algoritmo aprende automaticamente com os dados, os hiperparâmetros controlam a estrutura do modelo e a forma como o aprendizado acontece. Exemplos: a profundidade máxima de uma árvore, a quantidade de vizinhos em um modelo k-NN ou a taxa de aprendizado de uma rede neural.

O conjunto de **Teste** (*Test Set*) consiste em uma amostra de dados que permanece estritamente guardada e intocada durante toda a fase de desenvolvimento. Ele é utilizado uma única vez ao final do projeto para simular a aplicação do modelo no mundo real, fornecendo uma estimativa honesta e independente da sua precisão preditiva em dados não vistos.


![](https://github.com/alexlimatds/esp_introducao_am/blob/main/imagens/train_validation_test_sets.webp?raw=true) \
Fonte: https://medium.com/@masadeghi6/how-to-split-your-data-for-machine-learning-eae893a8799c

Também é comum a divisão de dados apenas em conjunto de treinamento e em conjunto de teste. No entanto, ajustar os hiperparâmetros com o conjunto de teste faz com que o desempenho final do modelo seja "contaminado" com dados de teste levando a uma estimativa excessivamente otimista da capacidade de generalização do modelo.
Em outras palavras, ao ajustar os hiperparâmetros com base no desempenho do conjunto de teste, você acaba fazendo um sobreajuste indireto (overfitting) em relação a esses dados. O conjunto de teste deixa de cumprir seu papel principal: servir como uma avaliação imparcial e totalmente independente com dados que o modelo nunca viu antes. O resultado final reportado parecerá excelente, mas o modelo provavelmente terá um desempenho inferior ao esperado ao encontrar novos dados no mundo real.


````python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
import numpy as np

# Carregando dataset da Iris
iris = load_iris()

# Separando o dataset em conjuntos de treino, validação e teste
# Como a função train_test_split faz separação apenas em dois conjuntos,
# devemos chamá-la duas vezes.
# Calculando tamanhos dos conjuntos
n_instances = iris.data.shape[0]  # quantidade de instâncias do dataset
test_size = int(0.2 * n_instances)
validation_size = int(0.2 * n_instances)

# separando em conjunto de treino e de teste
X_train, X_test, y_train, y_test = train_test_split(
    iris.data,
    iris.target,
    test_size=test_size,
    stratify=iris.target,
    random_state=42
)

# separando X_train em conjunto de treino e de validação
# repare que X_train e y_train são substituídos pelo novo conjunto
X_train, X_val, y_train, y_val = train_test_split(
    X_train,
    y_train,
    test_size=validation_size,
    stratify=y_train,
    random_state=42
)

print('Tamanho do dataset:', n_instances)
print('Tamanho do conjunto de treino:', X_train.shape[0])
print('Tamanho do conjunto de validação:', X_val.shape[0])
print('Tamanho do conjunto de teste: ', X_test.shape[0])
````

## Métricas de Avaliação para Classificação

Para quantificar a qualidade das predições feitas por modelos de classificação, utilizamos um conjunto de métricas estruturadas a partir das contagens de acertos e erros do algoritmo.

### Classificação Binária

Em problemas de classificação binária, o modelo classifica cada instância entre duas categorias: Classe Positiva e Classe Negativa. Para avaliar o desempenho dessas previsões e entender os tipos de acertos e erros cometidos pelo modelo, utilizamos métricas que são calculadas a partir dos seguintes componentes:

* **Verdadeiros Positivos ($VP$):** quantidade de instâncias da classe positiva que o modelo previu corretamente como positivas.
* **Falsos Positivos ($FP$):** quantidade de instâncias da classe negativa que o modelo erroneamente previu como positivas.
* **Falsos Negativos ($FN$):** quantidade de instâncias da classe positiva que o modelo erroneamente previu como negativas.
* **Verdadeiros Negativos ($VN$):** quantidade de instâncias da classe negativa que o modelo previu corretamente como negativas.

A partir desses dados, calculamos a matriz de confusão e as métricas que expressam o desempenho de um modelo: acurácia, precisão, sensibilidade, F-score

Para ilustrar o cálculo desses indicadores, considere o teste de triagem médica em um grupo de $100$ pacientes, no qual $30$ possuem uma doença (Positivos) e $70$ são saudáveis (Negativos). Suponha que um modelo de classificação produziu o seguinte resultado:

* $VP = 24$ (pacientes doentes identificados corretamente como doentes);
* $FN = 6$ (pacientes doentes erroneamente classificados como saudáveis);
* $FP = 14$ (pacientes saudáveis erroneamente classificados como doentes);
* $VN = 56$ (pacientes saudáveis identificados corretamente como saudáveis).

A **matriz de confusão** é uma tabela que cruza os rótulos reais das amostras (distribuídos no sentido vertical) com as predições geradas pelo modelo (distribuídas no sentido horizontal). Ela é constituída pelos quatro quadrantes ilustrados na figura abaixo.

![](https://github.com/alexlimatds/esp_introducao_am/blob/main/imagens/matriz_de_confusao.png?raw=true)

Assim, para o nosso exemplo teremos a seguinte matriz de confusão:

![](https://github.com/alexlimatds/esp_introducao_am/blob/main/imagens/matriz_de_confusao_2.png?raw=true)


A **Acurácia** (*Accuracy*) mede a fração de todas as previsões que foram corretas em relação ao total de casos:

$$\text{Acurácia} = \frac{VP + VN}{VP + VN + FP + FN}$$

Então, para o nosso exemplo teremos:

$$\text{Acurácia} = \frac{24 + 56}{24 + 56 + 14 + 6} = \frac{80}{100} = 0{,}80 \quad (80\%)$$

A Acurácia responde a seguinte pergunta: *De todas as previsões feitas, quantas o modelo acertou, tanto na classe positiva quanto na negativa?* A acurácia é intuitiva e útil quando as classes estão relativamente balanceadas e os dois tipos de erro têm custos parecidos. Porém, pode ser enganosa em dados desbalanceados: um modelo que prevê sempre a classe majoritária pode ter acurácia alta sem detectar a classe minoritária.

> Exemplo ilustrativo (filtro de spam): se, entre 1.000 e-mails, o modelo classifica corretamente 940, sua acurácia é 94%. Isso mostra o desempenho global, mas não revela se ele está deixando muitos spams passarem ou marcando e-mails legítimos como spam.

A **Precisão** (*Precision*) mede a proporção de instâncias realmente positivas entre todas aquelas que o modelo classificou como positivas (mas não fornece informação sobre as instâncias positivas classificadas como negativas). Ela indica a confiabilidade dos alarmes emitidos pelo modelo:

$$\text{Precisão} = \frac{VP}{VP + FP}$$

Para o nosso exemplo:

$$\text{Precisão} = \frac{24}{24 + 14} = \frac{24}{38} \approx 0{,}6316 \quad (63{,}16\%)$$

A Precisão responde a seguite pergunta: *Dos casos que o modelo marcou como positivos, quantos eram realmente positivos?* Ela penaliza falsos positivos e deve ser priorizada quando um alarme falso é custoso.

> Exemplo ilustrativo (filtro de spam): se o modelo marcar 100 e-mails como spam e 90 forem realmente spam, a precisão é 90%. Aqui, enviar um e-mail legítimo para a caixa de spam é um erro grave, então alta precisão é desejável.

A **Sensibilidade**, também conhecida como Revocação ou *Recall* (ou *True Positive Rate* - TPR), mede a proporção de instâncias realmente positivas que o modelo foi capaz de capturar com sucesso (mas não informa quantas instâncias negativas foram classificados incorretamente como positivas):

$$\text{Sensibilidade} = \frac{VP}{VP + FN}$$

Para o nosso exemplo:

$$\text{Sensibilidade} = \frac{24}{24 + 6} = \frac{24}{30} = 0{,}80 \quad (80\%)$$

A Sensibilidade responde a seguinte pergunta: *Dos casos realmente positivos, quantos o modelo conseguiu identificar?*. Ela penaliza falsos negativos e é prioritária quando deixar passar um caso positivo é perigoso ou caro.

> Exemplo ilustrativo (diagnóstico de câncer): se existem 100 pacientes doentes e o modelo identifica 85, o recall é 85%. Um falso negativo pode atrasar o tratamento, então Recall costuma ser mais crítico que acurácia ou precisão.

O **F-score** ($F_\beta$) combina Precisão e Sensibilidade em uma média harmônica ponderada, a qual penaliza com mais rigor os valores muito baixos de qualquer uma das duas métricas. O parâmetro $\beta$ permite ajustar o peso atribuído à Sensibilidade em relação à Precisão:

$$F_\beta = (1 + \beta^2) \cdot \frac{\text{Precisão} \cdot \text{Sensibilidade}}{(\beta^2 \cdot \text{Precisão}) + \text{Sensibilidade}}$$

Quando escolhemos $\beta = 1$, temos o score $F_1$, que atribui peso idêntico para a Precisão e a Sensibilidade. Aplicando ao exemplo, obtemos:

$$F_1 = (1 + 1^2) \cdot \frac{\text{Precisão} \cdot \text{Sensibilidade}}{\text{Precisão} + \text{Sensibilidade}} = 2 \cdot \frac{0{,}6316 \cdot 0{,}80}{0{,}6316 + 0{,}80} = \frac{1{,}0105}{1{,}4316} \approx 0{,}7058 \quad (70{,}58\%)$$

Quando $\beta = 2$, temos o score $F_2$, que atribui o dobro da importância à Sensibilidade em relação à Precisão. Isso é ideal para diagnósticos médicos ou detecção de fraudes, onde perder um caso positivo ($FN$) é muito mais perigoso do que um alarme falso ($FP$). Para o exemplo:

$$F_2 = (1 + 2^2) \cdot \frac{0{,}6316 \cdot 0{,}80}{(2^2 \cdot 0{,}6316) + 0{,}80} = 5 \cdot \frac{0{,}5053}{2{,}5264 + 0{,}80} = \frac{2{,}5264}{3{,}3264} \approx 0{,}7595 \quad (75{,}95\%)$$

Quando $\beta = 0{,}5$, temos o score $F_{0{,}5}$ que atribui maior peso à Precisão do que à Sensibilidade. Essa abordagem é apropriada para aplicações como filtros de *spam* em e-mails, onde classificar uma mensagem legítima e importante como *spam* ($FP$) traz prejuízos piores do que deixar um *spam* passar para a caixa de entrada ($FN$).

$$F_{0{,}5} = (1 + 0{,}5^2) \cdot \frac{0{,}6316 \cdot 0{,}80}{(0{,}5^2 \cdot 0{,}6316) + 0{,}80} = 1{,}25 \cdot \frac{0{,}5053}{0{,}1579 + 0{,}80} = \frac{0{,}6316}{0{,}9579} \approx 0{,}6594 \quad (65{,}94\%)$$

O F1-score é útil quando um equilíbrio entre Precisão e Sensibilidade é necessário. Por ser média harmônica, ele fica baixo se qualquer uma das duas métricas for baixa.

> Exemplo ilustrativo (detecção de fraude): um modelo pode ter Sensibilidade alta ao acusar muitas transações como fraudulentas, mas gerar muitos falsos alarmes; ou Precisão alta ao acusar pouquíssimas transações, deixando fraudes passarem. O F1 ajuda a comparar modelos que devem detectar bem as fraudes sem gerar excesso de falsos positivos.

**Resumo do exemplo**:

- Acurácia: 80%
- Precisão: 63,16%
- Sensibilidade: 80%
- F1-score: 70,58%
- Interpretação: o modelo acerta 80% das classificações no total, mas quando indica que alguém está doente, acerta apenas 63,16% das vezes. Além disso, o modelo identifica 80% das pessoas realmente doentes. Se o custo de não detectar a doença for alto, pode-se ajustar o modelo para aumentar a Sensibilidade, mesmo que a precisão caia um pouco.


### Classificação Multiclasse

Em problemas com mais de duas classes ($C > 2$), a matriz de confusão expande-se para uma matriz $C \times C$. Para estender métricas como Precisão, Sensibilidade e $F_1\text{-score}$ para esses cenários, adota-se a estratégia "Um-contra-Todos" (*One-vs-Rest*), calculando as métricas de forma individual para cada classe e, em seguida, aplicando um método de agregação global (macro ou ponderado).

Como exemplo ilustrativo, considere a seguinte matriz de confusão para um problema de classificação de 3 classes (A, B e C), onde as linhas representam as classes reais e as colunas as classes preditas pelo modelo:

```
                    Predição
                 A     B     C   | Total Real
               +-----+-----+-----+
             A | 30  |  5  |  5  |    40
Real         B |  3  | 40  |  7  |    50
             C |  2  |  3  | 25  |    30
               +-----+-----+-----+
Total Predito  | 35  | 48  | 37  |   120

```

Calculando as métricas por classe individualmente:

* **Classe A:**
  * $VP_A = 30$
  * $FP_A = 3 + 2 = 5$
  * $FN_A = 5 + 5 = 10$
  * $\text{Precisão}_A = \frac{30}{30 + 5} = \frac{30}{35} \approx 0{,}8571$
  * $\text{Sensibilidade}_A = \frac{30}{30 + 10} = \frac{30}{40} = 0{,}75$
  * $F1_{A} = 2 \cdot \frac{0{,}8571 \cdot 0{,}75}{0{,}8571 + 0{,}75} \approx 0{,}8$

* **Classe B:**
  * $VP_B = 40$
  * $FP_B = 5 + 3 = 8$
  * $FN_B = 3 + 7 = 10$
  * $\text{Precisão}_B = \frac{40}{40 + 8} = \frac{40}{48} \approx 0{,}8333$
  * $\text{Sensibilidade}_B = \frac{40}{40 + 10} = \frac{40}{50} = 0{,}8$
  * $F1_{B} = 2 \cdot \frac{0{,}8333 \cdot 0{,}8}{0{,}8333 + 0{,}8} \approx 0{,}8162$

* **Classe C:**
  * $VP_C = 25$
  * $FP_C = 5 + 7 = 12$
  * $FN_C = 2 + 3 = 5$
  * $\text{Precisão}_C = \frac{25}{25 + 12} = \frac{25}{37} \approx 0{,}6757$
  * $\text{Sensibilidade}_C = \frac{25}{25 + 5} = \frac{25}{30} \approx 0{,}8333$
  * $F1_{C} = 2 \cdot \frac{0{,}6757 \cdot 0{,}8333}{0{,}6757 + 0{,}8333} \approx 0{,}7463$


#### Estratégias de Agregação

A **Macro-média** (*Macro-average*) calcula a métrica de maneira isolada para cada classe e tira a média aritmética simples dos resultados. Ela atribui peso igual a todas as classes, independentemente do número de amostras que cada uma possui, sendo valiosa para destacar o desempenho do modelo em classes minoritárias:

$$\text{Precisão}_{\text{macro}} = \frac{\text{Precisão}_A + \text{Precisão}_B + \text{Precisão}_C}{3} = \frac{0{,}8571 + 0{,}8333 + 0{,}6757}{3} \approx 0{,}7887$$

$$\text{Sensibilidade}_{\text{macro}} = \frac{\text{Sensibilidade}_A + \text{Sensibilidade}_B + \text{Sensibilidade}_C}{3} = \frac{0{,}75 + 0{,}8 + 0{,}8333}{3} \approx 0{,}7944$$

$$F1_{\text{macro}} = \frac{F1_A + F1_B + F1_C}{3} = \frac{0{,}8 + 0{,}8162 + 0{,}7463}{3} \approx 0{,}7875$$

A **Média Ponderada** (*Weighted-average*) calcula a média das métricas de cada classe ponderando-as pelo suporte (quantidade real de instâncias) de cada uma no conjunto de dados. Ela reflete o desempenho proporcional à presença de cada classe na amostra:

$$\text{Precisão}_{\text{weighted}} = \frac{|A| \cdot \text{Precisão}_A + |B| \cdot \text{Precisão}_B + |C| \cdot \text{Precisão}_C}{|A|+|B|+|C|} = \frac{(40 \cdot 0{,}8571) + (50 \cdot 0{,}8333) + (30 \cdot 0{,}6757)}{120} = \frac{34{,}284 + 41{,}665 + 20{,}271}{120} = \frac{96{,}22}{120} \approx 0{,}8018$$

$$\text{Sensibilidade}_{\text{weighted}} = \frac{(40 \cdot 0{,}75) + (50 \cdot 0{,}8) + (30 \cdot 0{,}8333)}{120} = \frac{30 + 40 + 24{,}999}{120} = \frac{94{,}999}{120} \approx 0{,}7916$$

$$F1_{\text{weighted}} = \frac{(40 \cdot 0{,}8) + (50 \cdot 0{,}8162) + (30 \cdot 0{,}7463)}{120} = \frac{32 + 40{,}81 + 22{,}389}{120} = \frac{95{,}199}{120} \approx 0{,}79331$$

````python
# Treinando um classificador e gerando as métricas
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report, precision_recall_fscore_support
import matplotlib.pyplot as plt

knn_classifier = KNeighborsClassifier(n_neighbors=3)
knn_classifier.fit(X_train, y_train)
y_pred = knn_classifier.predict(X_val)

# matriz de confusão
conf_matrix = confusion_matrix(y_val, y_pred)
cm_disp = ConfusionMatrixDisplay(
    confusion_matrix=conf_matrix,
    display_labels=iris.target_names
)
cm_disp.plot()
plt.title('Confusion Matrix')
plt.show()

# relatório de métricas
print('\nRelatório de Métricas:')
report = classification_report(y_val, y_pred, target_names=iris.target_names)
print(report)

# computando as métricas
p_macro, r_macro, f_macro, _ = precision_recall_fscore_support(y_val, y_pred, average='macro')
print('Precision macro:', p_macro)
print('Recall macro:', r_macro)
print('F-score macro:', f_macro)
````

## Formas de Amostragem de Dados

A forma como dividimos e amostramos nossos dados determina a confiabilidade e a estabilidade da estimativa de desempenho do modelo.

### Método Holdout

O método Holdout consiste na divisão mais simples do conjunto de dados em subconjuntos fixos (como 70% para treinamento e 30% para teste).

![](https://github.com/alexlimatds/esp_introducao_am/blob/main/imagens/holdout.png?raw=true) \
Fonte: https://dhavalpatel2101992.wordpress.com/2021/05/21/kaggle-titanic-dataset-cleaning-split-data-into-train-validation-and-test-set/

A principal vantagem do Holdout é a sua rapidez computacional e facilidade de implementação. No entanto, sua desvantagem reside na alta sensibilidade à partição sorteada (variância de amostragem). Se a divisão isolar amostras atípicas no conjunto de teste, o resultado não refletirá o desempenho real do algoritmo no domínio completo.




### Validação Cruzada ($k$-fold Cross-Validation)

A validação cruzada $k$-fold é uma estratégia robusta desenvolvida para mitigar o viés associado ao particionamento do método Holdout. No $k$-fold, o conjunto de dados completo é dividido aleatoriamente em $k$ partes de tamanhos iguais, chamadas de *folds*.

O processo executa $k$ iterações sequenciais. Em cada iteração $i$, o $i$-ésimo *fold* é separado exclusivamente como conjunto de teste, enquanto os outros $k-1$ *folds* são unidos para compor o conjunto de treinamento. O modelo é treinado e avaliado, gerando uma métrica de desempenho individual. Ao final do ciclo de $k$ execuções, o desempenho final reportado é a média aritmética do desempenho obtido através de todos os $k$ testes:

![](https://github.com/alexlimatds/esp_introducao_am/blob/main/imagens/k_fold_cross_validation.png?raw=true) \
Fonte: https://www.analyticsvidhya.com/blog/2021/05/importance-of-cross-validation-are-evaluation-metrics-enough/


Essa abordagem garante que todas as amostras do conjunto de dados sejam utilizadas exatamente uma vez como dados de teste e $k-1$ vezes como dados de treinamento, reduzindo drasticamente o risco de viés na estimativa. Os valores mais comuns adotados na prática para $k$ são 5 e 10.


````python
# A forma mais prática de realizar uma validação cruzada no
# Scikit-learn é usando a função cross_validate
from sklearn.model_selection import cross_validate
import pandas as pd

classificador = KNeighborsClassifier(n_neighbors=3)
chosen_metrics = ['accuracy', 'precision_macro', 'recall_macro', 'f1_macro']
cv_results = cross_validate(
    classificador,
    iris.data,
    iris.target,
    cv=5,  # quantidade de partições
    scoring=chosen_metrics
)
results_df = pd.DataFrame(cv_results)
display(results_df)
print("\nMédias:")
print(results_df.mean())
````

### Leave-One-Out Cross-Validation (LOOCV)

O *Leave-One-Out* é o caso limite da validação cruzada $k$-fold, em que o número de partições $k$ é configurado para ser igual ao número total de instâncias $N$ presentes no conjunto de dados ($k = N$).

Em cada iteração, exatamente **uma única amostra** é reservada para teste, e todas as $N-1$ amostras restantes são utilizadas para treinar o modelo. Esse procedimento é repetido $N$ vezes, de modo que cada ponto do conjunto de dados seja avaliado isoladamente.

A principal vantagem do LOOCV é o aproveitamento máximo dos dados disponíveis para treinamento em cada passo, sendo deterministicamente estável (não depende de sorteios aleatórios). É especialmente recomendado para amostras extremamente pequenas. Contudo, sua aplicação torna-se inviável em conjuntos de dados de médio e grande porte devido ao elevado custo computacional, já que exige o treinamento completo do algoritmo de aprendizado $N$ vezes consecutivas.

````python
# Para o Levae-one-out usamos LeaveOneOut
from sklearn.model_selection import LeaveOneOut

classificador = KNeighborsClassifier(n_neighbors=3)
loo = LeaveOneOut()
scores = []
for train_idx, test_idx in loo.split(iris.data):  # a função split retorna os índices aos invés das instâncias
    classificador.fit(iris.data[train_idx], iris.target[train_idx])
    y_pred = classificador.predict(iris.data[test_idx])
    y_val = iris.target[test_idx]
    p_macro, r_macro, f_macro, _ = precision_recall_fscore_support(y_val, y_pred, average='macro')
    scores.append([p_macro, r_macro, f_macro])

scores = np.array(scores)

results_df = pd.DataFrame(scores, columns=['Precision (macro)', 'Recall (macro)', 'F1 (macro)'])
display(results_df)
print("\nMédias:")
print(results_df.mean())
````
